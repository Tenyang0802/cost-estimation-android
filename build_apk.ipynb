## 📱 成本估算软件 - Android APK一键打包

此笔记本自动将Kivy代码打包为Android APK

**使用方法：** 菜单栏 → 运行 → 全部运行（约3-5分钟）

```
# 1. 安装Buildozer依赖
!pip install --quiet buildozer cython
!apt-get update -qq
!apt-get install -y -qq git zip unzip openjdk-17-jdk python3-pip autoconf libtool pkg-config zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev 2>/dev/null
print('✓ 环境就绪')
```

```
# 2. 创建项目目录并写入代码
import os, json

PROJECT_DIR = '/content/cost-estimation-android'
os.makedirs(PROJECT_DIR, exist_ok=True)

# ========== 写入 data_manager.py ==========
dm_code = open('/tmp/dm_code.py', 'w')
# (代码在上一步已写入会从本地源码读取)
print('请将以下文件上传到 content/ 目录：')
print('1. main.py')
print('2. data_manager.py')
print('3. buildozer.spec')
print('4. cost_data.json')
```

```
# 3. 上传代码文件（运行后点左侧文件夹图标上传）
from google.colab import files
print('请点击左侧文件夹图标，将以下文件拖入 /content/ 目录：')
print('  - main.py')
print('  - data_manager.py')
print('  - buildozer.spec')
print('  - cost_data.json')
print()
print('或者直接运行下面代码从GitHub下载（如果已上传则跳过）')

# 检查文件是否存在
import os
files_needed = ['main.py', 'data_manager.py', 'buildozer.spec', 'cost_data.json']
missing = [f for f in files_needed if not os.path.exists(f'/content/{f}')]
if missing:
    print(f'缺少的文件: {missing}')
    print('请点击左侧文件夹图标上传')
else:
    print('✓ 所有文件已就绪！')
```

```
# 4. 整理项目目录
import shutil, os

src_dir = '/content'
dst_dir = PROJECT_DIR

for f in ['main.py', 'data_manager.py', 'buildozer.spec', 'cost_data.json']:
    src = os.path.join(src_dir, f)
    if os.path.exists(src):
        shutil.copy2(src, os.path.join(dst_dir, f))
        print(f'✓ {f}')
    else:
        print(f'✗ {f} 未找到，请上传')

os.chdir(dst_dir)
print(f'\n项目目录: {dst_dir}')
!ls -la
```

```
# 5. 🚀 开始打包APK（约3-5分钟）
import os
os.chdir(PROJECT_DIR)
print('正在打包APK，请耐心等待...')
!buildozer android debug 2>&1 | tail -30
```

```
# 6. 下载APK
from google.colab import files
import glob

apk_files = glob.glob(f'{PROJECT_DIR}/bin/*.apk')
if apk_files:
    print(f'找到APK: {apk_files[0]}')
    files.download(apk_files[0])
else:
    print('APK未找到，请检查打包日志')
```
